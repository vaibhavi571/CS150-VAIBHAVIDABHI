## Task 1: Weekend Activities as Sets
**A**. Let the user input activities for Saturday and store them in a set called actSat. "Done" stops the input.

```python
actSat = set()
act = input('Enter an activity from last Saturday ("Done" to stop):')
while act != "Done":
    actSat.add(act)
    act = input('Enter an activity from last Saturday ("Done" to stop):')
```

**B**. Do the same for Sunday and store the activities in a set called actSun.

```python
actSun = set()
act = input('Enter an activity from last Sunday ("Done" to stop):')
while act != "Done":
    actSun.add(act)
    act = input('Enter an activity from last Sunday ("Done" to stop):')
```

**C**. Print both sets

```python
print(actSat)
print(actSun)
```

**D** .Print the activities done on both days

```python
print(actSat.intersection(actSun))
```

**E**. Print the activities done on at least one of the days

```python
print(actSat.union(actSun))
```

## Task 2: Triangle Calculations
Write a function triangle_solve(leg1, leg2) that returns the hypotenuse, perimeter, and area of a right triangle as a tuple.

```python
import math
leg1 = int(input("Enter the length of leg 1:"))
leg2 = int(input("Enter the length of leg 2:"))

def triangle_solve(leg1, leg2):
    hypotenuse = math.sqrt(leg1**2 + leg2**2)
    perimeter = leg1 + leg2 + hypotenuse
    area = leg1 * leg2 / 2
    return hypotenuse, perimeter, area

print(triangle_solve(leg1,leg2))
```

## Task 3: Calculator Class
Create a class named Calculator with add, subtract, multiply, divide, and power methods. It should keep track of prev_result and total_calculations, and have an info() method that prints them.

```python
class Calculator():
    def __init__(self):
        self.prev_result = 0
        self.total_calculations = 0

    def add(self, a, b):
        self.prev_result = a + b
        self.total_calculations += 1
        return self.prev_result

    def subtract(self, a, b):
        self.prev_result = a - b
        self.total_calculations += 1
        return self.prev_result

    def multiply(self, a, b):
        self.prev_result = a * b
        self.total_calculations += 1
        return self.prev_result

    def divide(self, a, b):
        self.prev_result = a / b
        self.total_calculations += 1
        return self.prev_result

    def power(self, a, b):
        self.prev_result = a ** b
        self.total_calculations += 1
        return self.prev_result

    def info(self):
        print(f"Previous result: {self.prev_result}")
        print(f"Total calculations: {self.total_calculations}")

calc = Calculator()
print(calc.add(5, 3))
print(calc.subtract(10, 4))
print(calc.multiply(6, 7))
print(calc.divide(20, 4))
print(calc.power(2, 5))
calc.info()
```
