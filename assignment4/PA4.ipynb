## Task 1: Input Positive Integer and Factors (Part A)

## Task 1: Input Positive Integer and Factors (Part B)

```python
NUM=int(input("please enter positive integer:"))
numcount=0
for i in range(1,100):
    if NUM % i== 0:
        numcount+=1
if numcount ==2:
    print("This number is prime")
else:
    print("This number is not prime")
    
```

## Task 3: Temperature Summaries 

```python
temps = {'January': [43, 26], 'February': [48, 29], 'March': [57, 37],
         'April': [69, 47], 'May': [77, 58], 'June': [85, 66],
         'July': [89, 70], 'August': [88, 69], 'September': [82, 62],
         'October': [70, 50], 'November': [57, 38], 'December': [47, 31]}
month = input("Enter a month: I will give you the average high and low temperatures in Louisville")
if month in temps:
        print("The average high temperature in Louisville for", month, "is", temps[month][0],
              "and the average low is", temps[month][1])
else:
        print("Not a valid month")

```

## Task 4: Inventory Dictionary and Prices 

```python
inventory = {'hammer': [15, 12.49], 'wrench': [20, 7.99], 'screwdriver': [40, 3.99],
             'nailgun': [10, 36.59], 'padlock': [25, 5.99], 'roll_of_duct_tape': [30, 4.95]}
print("1. Look up an item")
print("2. Show items under a budget")
print("3. Total quantity of items")
print("4. Total price of all items")
choice = int(input("Choose an option (1-4): "))
if choice == 1:
    item = input("Enter an item name: ")
    if item in inventory:
        print("Quantity:", inventory[item][0], " Price:", inventory[item][1])
    else:
        print("Sorry, that item is not available.")
elif choice == 2:
    budget = int(input("Enter your maximum budget: "))
    for item in inventory:
        if inventory[item][1] < budget:
            print(item, inventory[item][1])
elif choice == 3:
    total = 0
    for item in inventory:
        total = total + inventory[item][0]
    print("Total quantity of items:", total)

elif choice == 4:
    total = 0
    for item in inventory:
             total = total + inventory[item][0] * inventory[item][1]
    print("Total price of all items:", total)
else:
    print("invalid choice")
```
