## Task 1: Types of Triangles

Write a program that takes user input representing three sides of a triangle. Determine whether the sides represent an equilateral, isosceles, or scalene triangle.

```python
side1 = float(input("Enter the first side of the triangle: "))
side2 = float(input("Enter the second side of the triangle: "))
side3 = float(input("Enter the third side of the triangle: "))

if side1 == side2 == side3:
    print("The triangle is equilateral.")
elif side1 == side2 or side1 == side3 or side2 == side3:
    print("The triangle is isosceles.")
else:
    print("The triangle is scalene.")
```

## Task 2: Student Grades and At-Risk Status

Read five student grade averages into an initially empty list called `student_grades`. Calculate and print the class average, minimum grade, maximum grade, and number of students with a grade below 65.

```python
student_grades = []

for i in range(5):
    grade = float(input("Enter a student grade: "))
    student_grades.append(grade)

class_average = sum(student_grades) / len(student_grades)
minimum_grade = min(student_grades)
maximum_grade = max(student_grades)

print("Class average:", class_average)
print("Minimum grade:", minimum_grade)
print("Maximum grade:", maximum_grade)

at_risk = 0

for grade in student_grades:
    if grade < 65:
        at_risk += 1

print(at_risk, "students are at risk in the course.")
```

## Task 3: Ranges and Multiples

Create a list called `mult_of_2` containing all multiples of 2 between 1 and 100, a list called `mult_of_3` containing all multiples of 3, a list called `mult_of_both` containing numbers present in both lists, and a list called `mult_of_either` containing numbers present in either list.

```python
mult_of_2 = []
mult_of_3 = []

for number in range(1, 101):
    if number % 2 == 0:
        mult_of_2.append(number)

for number in range(1, 101):
    if number % 3 == 0:
        mult_of_3.append(number)

mult_of_both = []

for number in mult_of_2:
    if number in mult_of_3:
        mult_of_both.append(number)

mult_of_either = []

for number in range(1, 101):
    if number in mult_of_2 or number in mult_of_3:
        mult_of_either.append(number)

print("Multiples of 2:")
print(mult_of_2)

print("\nMultiples of 3:")
print(mult_of_3)

print("\nMultiples of both:")
print(mult_of_both)

print("\nMultiples of either:")
print(mult_of_either)
```

## Task 4: Shopping List

Starting with an initially empty list named `shopping_list`, allow the user to enter up to 7 items. The user can enter `done` to stop early. Then print the total number of items, the list in reverse order, the list in alphabetical order, and any item that appears at least twice.

```python
shopping_list = []

for i in range(7):
    item = input("Enter an item or type 'done' to finish: ")

    if item.lower() == "done":
        break

    shopping_list.append(item)

print("Total number of items:", len(shopping_list))

print("Shopping list in reverse order:")
print(shopping_list[::-1])

print("Shopping list in alphabetical order:")
print(sorted(shopping_list))

print("Items that appear more than once:")

duplicates = []

for item in shopping_list:
    if shopping_list.count(item) >= 2 and item not in duplicates:
        duplicates.append(item)

if len(duplicates) == 0:
    print("No items appear more than once.")
else:
    for item in duplicates:
        print(item)
```
